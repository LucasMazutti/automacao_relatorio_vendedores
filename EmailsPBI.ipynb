```python
import pandas as pd
import smtplib
import os

from email.message import EmailMessage
from getpass import getpass

# Caminho e nome do arquivo Excel
arquivo = r"\\VCICWBSRVFLS02\Unidade_G\TI\POWER BI OFICIAL\RELATORIOS\RelatorioNFs.xlsx"

# Exporta o DataFrame para Excel
df.to_excel(
    arquivo,
    index=False,
    sheet_name="fato_Relatorio"
)

print("Arquivo Excel gerado com sucesso!")
```

```python
# Servidor SMTP da empresa
SMTP_SERVER = "mail.vitao.com.br"

# Porta SMTP (informada pela TI)
SMTP_PORT = 587

# Conta corporativa que fará o envio
EMAIL_REMETENTE = "bi@vitao.com.br"

# Senha da conta (será solicitada sem aparecer na tela)
EMAIL_SENHA = getpass("Digite a senha do e-mail: ")

# Destinatário do teste
EMAIL_DESTINATARIO = "lucas.mazutti@vitao.com.br"
```

```python
# Criação da mensagem
msg = EmailMessage()

msg["Subject"] = "Teste - Relatório de Faturamento"
msg["From"] = EMAIL_REMETENTE
msg["To"] = EMAIL_DESTINATARIO

# Corpo do e-mail
msg.set_content("""
Bom dia!

Segue em anexo o relatório de faturamento para teste.

Atenciosamente,
Equipe BI
""")

# Anexar arquivo Excel
with open(arquivo, "rb") as anexo:
    msg.add_attachment(
        anexo.read(),
        maintype="application",
        subtype="vnd.openxmlformats-officedocument.spreadsheetml.sheet",
        filename=os.path.basename(arquivo)
    )
```

```python
try:
    with smtplib.SMTP(SMTP_SERVER, SMTP_PORT) as servidor:

        # Inicia a conexão criptografada
        servidor.starttls()

        # Autentica a conta de e-mail
        servidor.login(EMAIL_REMETENTE, EMAIL_SENHA)

        # Envia a mensagem com o Excel anexado
        servidor.send_message(msg)

    print("E-mail enviado com sucesso!")

except Exception as erro:
    print("Erro ao enviar e-mail:", erro)
```
